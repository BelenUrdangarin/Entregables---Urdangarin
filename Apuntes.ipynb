Dentro de ML hay dos tipos, apredizaje supervisado y no supervisado. El primero aprende utilizando datos etiquetados, mientras que el segundo no.

El random forest es de regresion y clasificación.
Cada uno de los metodos tiene sus parametros de ajuste.

Los hiperparametros son configuraciones previas al entrenamientos que se ajustan para encontrar los mejores modelos para nuestros datos, lo que se usa es la validacion cruzada para elegir los mejores.
El underfitting es cuando el modelo es demasiado simple y no captura la estructura de datos y el overfitting es cuando el modelo se ajusta demasiado a los datos de entrenamiento y tiene un bajo rendimiento en los datos.

Diferentes estrategias para evitar el over o under:
- Penalizar los coeficientes del modelo que se hacen que se ajuste demasiado
- Limitar la complejidad del modelo, simplificarlo, evitar patrones que no son relevantes.
- Aumentar los datos con rotaciones o recortes, aplicar diferentes transformaciones
- Usar la validacion cruzada
- Eliminacion de caracteristicas irrelevantes
- Uso de drop out, redes neuronales para evitar la dependencia de patrones

La claves es encontrar el **equilibrio adecuado entre el sesgo y varianza** sin perder la progresión.
